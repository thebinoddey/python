## Handling missing values

```python
data.isnull() #true for nul or NaN values
```

```python
data.isnull().any(axis=1) #any value from columns
```

```python
data.isnull().Salary #isnull for column Salary
```

```python
data.isnull().sum() #total null values wrt each columns
```

```python
#data.fillna(0) ---> #fills null with 'zero'
```

### Fill missing with mean

```python
data['Salary_fillNa'] = data['Salary'].fillna(data['Salary'].mean()) #we created new Salary_fillNa to fill null values of Salary in data with the mean of salary.
print(data['Salary_fillNa'])
```

## Renaming Column

```python
data = data.rename(columns={'Employee_ID':'EID'})
print(data.head(5))
```

## Change data type

```python
data.dtypes
```

```python
data['Salary_new'] = data['Salary'].astype('Int64') #Int64 for null values (capital I)
print(data['Salary_new'].dtype)
```

##  Data Aggregating & Grouping

```python
data.head(5)
```

```python
grouped_mean = data.groupby('Age')['Performance_Score'].mean().astype(int) #mean perf score grouped by age
print(grouped_mean)
```

```python
grouped_sum = data.groupby('Age')['Performance_Score'].sum()
print(grouped_sum)
```

```python
##aggregate function
grouped_agg = data.groupby("Age")["Performance_Score"].agg(['mean', 'sum', 'count'])
print(grouped_agg)
```
