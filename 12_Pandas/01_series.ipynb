```python
import pandas as pd
import numpy as np
```

# Series

 It is a **One-dimensionsal** labelled array.

```python
s = pd.Series(np.arange(1,23,4))
print(s)
```

```python
s = pd.Series(np.random.randn(5), index=["a", "b", "c", "d", "e"])
print(s)
```

```python
s.index
```

```python
d = {"b": 1, "a": 0, "c": 2}
pd.Series(d)
```

```python
pd.Series(d, index=["b", "c", "d", "a"])
# NaN (not a number) is the standard missing data marker used in pandas.
```

```python
pd.Series(5.0, index=["a", "b", "c", "d", "e"])
```

```python
data = [10, 20, 30, 40]
index = ['a', 'b', 'c', 'd']
pd.Series(data, index = index)
```
