# DataFrame

DataFrame is a 2D labeled data structure with columns of potentially different data types. It is like a spreadsheet

## DataFrame from dict of series or dicts

```python
d = {
    "one": pd.Series([1,2,3]),
    "two": pd.Series([2,3,4]),
    "three": pd.Series([3.12, 4.02, 5.23 ])
}
```

```python
df = pd.DataFrame(d)
print(df)
```

```python
d = {
    "one": pd.Series([1,2,3]),
    "two": pd.Series([2,3,4]),
    "three": pd.Series([3.12, 4.02, 5.23 ])
}
df = pd.DataFrame(d, index=[1,2])
print(df)
```

```python
df.index
```

```python
df.columns
```

## From Dict of ndarrays/lists

```python
d = {"one": [1.0, 2.0, 3.0, 4.0], "two": [4.0, 3.0, 2.0, 1.0]}
print(d)

```

```python
df = pd.DataFrame(d)
print(df)
```

## From structures or record array

This is handled identically to a dict of arrays.

```python
data = np.zeros((2,), dtype=[("A", "i4"), ("B", "f4"), ("C", "S10")])
data_df = pd.DataFrame(data)
print(data_df)
```

```python
data[:] = [(1, 2.0, "Hello"), (2, 3.0, "World")]
pd.DataFrame(data)
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 22, 28],
    "City": ["New York", "Chicago", "Boston", "Seattle"],
    "Salary": [55000, 62000, 48000, 71000]
}

people = pd.DataFrame(data)

print(people)
```
