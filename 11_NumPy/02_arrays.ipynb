# Creating a Basic Array

```python
import numpy as np
```

## 1D Array

```python
a = np.array([1, 2, 3])  # Create a 1D array
print(a)
print(a[0]) #Print first element of array
```

```python
np.zeros(2)
```

```python
b = np.zeros(3)
print(b)
```

```python
np.empty(3)
```

```python
np.ones(3)
```

## 2D Array

```python
b = np.array([[1,2,3],[4,5,6]])
print(b)
print(b[0,0])#Print first element (0 th row, 0th col) of 2D array
print(b[1,1])#Print element at (1,1) of 2D array
```

```python
np.zeros((2,3))
```

```python
z = np.zeros((3,4))
print(z)
```

```python
np.identity(2)
```

```python
np.eye(2)
```

## np.arange()

```python
c = np.arange(4)
print(c)
```

```python
d = np.arange(2,9,2)  #Start at 2, end before 9, step by 2
print(d)
```

```python
e = np.arange(5,16,3)  #Start at 5, end before 16, step by 3
print(e)
```

## np.linspace()

```python
np.linspace(0, 10, 5) #Generate 5 values spaced linearly between 0 and 10
```

```python
np.linspace(0,10,6) #Generate 6 values spaced linearly between 0 and 10
```

## Specifying your data type

```python
x = np.ones(2, dtype=int)
print(x)
```

```python
x = np.array([1,2.2,2], dtype=int)
print(x)
```
