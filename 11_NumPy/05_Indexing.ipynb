# Selecting Values

```python
import numpy as np
```

```python
a = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(a[a<5]) #print values less than 5
```

```python
five_up = (a>5)
print(a[five_up]) #print values greater than 5  
```

```python
print(a[a%2 == 0]) #print even values only
```

```python
five_up = (a > 5) | (a == 5)
print(five_up) #boolean array where values greater than or equal to 5 are True
```

## np.nonzero() to print the indices of elements that are

```python
b = np.nonzero(a<5)
print(b)
```
