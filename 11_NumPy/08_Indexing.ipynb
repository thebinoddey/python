## Indexing With Array of Indices

```python
import numpy as np
```

```python
a = np.arange(12)**2 #the first 12 square numbers
i = np.array([1,3,5,7,9]) #an array of indices3
```

```python
a[i] # the elements of "a" at the positions specified by "i"
```

```python
j = np.array( [ [ 3, 4], [ 9, 7 ] ] ) # a bidimensional array of indices
a[j]
```

##### When indexed array "a" is multidimensional, a single array of indices refer to the first dimension of "a"
