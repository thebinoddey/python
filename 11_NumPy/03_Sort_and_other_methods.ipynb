# Sorting array

```python
import numpy as np
```

```python
arr = np.array([1, 2, 23, 15, 5])
print(arr)
```

```python
np.sort(arr)
```

```python
np.argsort(arr) #Get the indices that would sort the array
```

# Concatenation

```python
a = np.array([1, 2, 3]) 
b = np.array([4, 5, 6])
np.concatenate((a, b))  #Concatenate arrays a and b
```

# Shape & Size

```python
ar = np.array([3, 1, 2])
ar.ndim  #Get the number of dimensions of the array
```

```python
ar.size #Get the total number of elements in the array
```

```python
ar.shape #Get the shape of the array
```

# arr.reshape()

```python
a = np.arange(6)  #Create an array with values from 1 to 10 with a step of 2
print(a)
```

```python
b = a.reshape(3, 2) 
print(b)
```

```python
np.reshape(a, shape=(1,6), order='c')
```
