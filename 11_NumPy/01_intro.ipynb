# What is NumPy?

- NumPy is the fundamental package for scientific computing in Python.
- It is a Python library that provides a multidimensional array object, various derived objects (such as masked arrays and matrices), and an assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, discrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.

```python
import numpy as np #Import Numpy after downloading as np (where np is an alias)
```

```python
a = np.array([1, 2, 3])  # Create a 1D array
print(a)
print(a[0]) #Print first element of array
```

```python
b = np.array([[1,2,3],[4,5,6]])
print(b)
print(b[0,0])#Print first element (0 th row, 0th col) of 2D array
print(b[1,1])#Print element at (1,1) of 2D array
```

```python
comp = np.array([[1,2],[3,4.6+ 25j]],dtype = complex)
```

```python
print(comp)
```

## Basic Operations

```python
a = np.array([20, 30, 40, 50])
b = np.arange(4)
print(a)
print(b)
```

```python
#Addition 
a + b

#subtraction
a - b

#product
a * b

#divide
a / (b + 1)
```

```python
#Power
a ** b
```

```python
a < 35
```

```python
A = np.array([[1, 1], [5, 1]])
B = np.array([[2, 1], [3, 4]])
```

```python
A * B #elementwise product
```

```python
A @ B #matrix product
```
